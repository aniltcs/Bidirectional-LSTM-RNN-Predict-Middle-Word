```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Bidirectional, LSTM, Dense
from sklearn.preprocessing import LabelEncoder
```

```python
texts = [
    "Anil eats something spicy in Bangalore",
    "Ravi enjoys something for breakfast in Chennai",
    "Sita loves to eat something hot in Delhi",
    "Geeta orders something delicious in Hyderabad",
    "Mohan prefers something street-style in Mumbai",
    "Asha eats something traditional in Kochi",
    "Rahul tries something fishy in Kolkata",
    "Meena enjoys something Rajasthani in Jaipur",
    "Kiran picks something tangy in Surat",
    "Divya eats something sweet in Lucknow",
    "Arjun likes something crispy in Indore",
    "Neha tries something cheesy in Pune",
    "Kavita orders something fried in Amritsar",
    "Suraj eats something coconut-based in Goa",
    "Priya enjoys something steamed in Madurai",
    "Manish loves something masala-rich in Nagpur",
    "Tina prefers something fermented in Manipur",
    "Raj eats something grilled in Chandigarh",
    "Sneha orders something stuffed in Agra",
    "Vikram eats something buttery in Ludhiana"
]

labels = [
    "Dosa",         
    "Idli",         
    "Paratha",      
    "Biryani",      
    "Vada Pav",     
    "Appam",        
    "Fish Curry",   
    "Dal Baati",    
    "Dhokla",       
    "Gulab Jamun",  
    "Kachori",      
    "Cheese Dosa",  
    "Amritsari Kulcha",  
    "Prawn Curry",       
    "Idiyappam",         
    "Misal Pav",         
    "Fermented Bamboo",  
    "Tandoori Chicken",  
    "Stuffed Paratha",   
    "Butter Chicken"     
]
```

```python
##Text Tokenization and label encoding

# Replace "something" for token consistency
texts = [text.replace('_', 'something') for text in texts]

# Tokenize
tokenizer = Tokenizer()
tokenizer.fit_on_texts(texts)
sequences = tokenizer.texts_to_sequences(texts)
vocab_size = len(tokenizer.word_index) + 1

max_len = max(len(seq) for seq in sequences)
X = pad_sequences(sequences, maxlen=max_len, padding='post')

# Encode labels
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(labels)
num_classes = len(label_encoder.classes_)
print(y)
print(num_classes)
```

```python
## Bidirectional Model
model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=32, input_length=max_len),
    Bidirectional(LSTM(64)),
    Dense(num_classes, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.summary()
```

```python
model.fit(X, y, epochs=50, batch_size=4, verbose=1)
```

```python
## Save the model,tokenizer and encoder

import pickle

# Save the model
model.save("food_predictor_model.h5")

# Save tokenizer
with open("tokenizer.pkl", "wb") as f:
    pickle.dump(tokenizer, f)

# Save label encoder
with open("label_encoder.pkl", "wb") as f:
    pickle.dump(label_encoder, f)
```

```python
## Function to predict word

def predict_missing_food(sentence):
    sentence = sentence.replace('_', 'something')
    seq = tokenizer.texts_to_sequences([sentence])
    padded = pad_sequences(seq, maxlen=max_len, padding='post')
    pred = model.predict(padded)
    index = np.argmax(pred)
    return label_encoder.inverse_transform([index])[0]
```

```python
# Seen input
print(predict_missing_food("Meena enjoys something Rajasthani in Jaipur"))

# Unseen but similar context
print(predict_missing_food("Ankit loves something buttery in Ludhiana"))

print(predict_missing_food("Ritika enjoys something tangy in Gujarat"))

print(predict_missing_food("Vivek eats something fried in Punjab"))

print(predict_missing_food("Karan tries something cheesy in Pune"))
```
